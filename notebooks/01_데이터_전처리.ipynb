# 스마트폰 가격 데이터 전처리
ZIP의 원본 전처리 노트북을 저장소 경로에 맞게 정리한 공개용 사본입니다.


```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency, kruskal
```

```python
df = pd.read_csv('../data/raw/train.csv')
print(df.shape)
display(df.head())
print(df.isna().sum())
print(df['price_range'].value_counts().sort_index())
```

```python
df['sc_total'] = df['sc_h'] * df['sc_w']
df['px_total'] = df['px_height'] * df['px_width']
binary_cols = ['blue','dual_sim','four_g','three_g','touch_screen','wifi']
for col in binary_cols:
    table = pd.crosstab(df[col], df['price_range'])
    print(col, chi2_contingency(table)[1])
numeric_cols = [c for c in df.columns if c not in binary_cols + ['price_range']]
for col in numeric_cols:
    groups = [g[col].dropna().to_numpy() for _, g in df.groupby('price_range')]
    print(col, kruskal(*groups).pvalue)
```

```python
selected = ['battery_power','int_memory','mobile_wt','ram','px_total','price_range']
df[selected].to_csv('../data/processed/train_preprocessed.csv', index=False)
```
